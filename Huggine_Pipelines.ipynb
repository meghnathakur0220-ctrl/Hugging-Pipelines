```python
from transformers import pipeline
from PIL import Image
import requests
```

Text Pipeline **Demonstration**

Task: Sentiment Analysis

```python
from transformers import pipeline

classifier = pipeline("sentiment-analysis")

result = classifier("I really enjoyed learning Hugging Face pipelines!")

print(result)
```

```python
classifier("This movie was boring.")
```

Image Pipeline **Demonstration**

Task: Image Classification

```python
url = "https://huggingface.co/datasets/huggingface/documentation-images/resolve/main/beignets-task-guide.png"

image = Image.open(requests.get(url, stream=True).raw)
```

```python
from transformers import pipeline

classifier = pipeline("image-classification")

result = classifier(image)

print(result)
```

Audio Pipeline **Demonstration**

Task: Speech Recognition

```python
!pip install datasets
```

```python
from datasets import load_dataset
from transformers import pipeline

dataset = load_dataset("hf-internal-testing/librispeech_asr_demo", "clean", split="validation")

audio = dataset[0]["audio"]

transcriber = pipeline("automatic-speech-recognition")

result = transcriber(audio)

print(result)
```

Video Pipeline **Demonstration**

Hugging Face does not have a simple "video-classification" pipeline like text or image pipelines. Instead, you can use a video classification model.

```python
!pip install opencv-python
```

```python
import cv2

video = cv2.VideoCapture("sample.mp4")

print("Video loaded successfully")
```

```python
from transformers import pipeline

classifier = pipeline(
    task="video-classification",
    model="MCG-NJU/videomae-base"
)
```

```python
result = classifier("sample.mp4")
print(result)
```
